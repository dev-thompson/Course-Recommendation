```python
import pandas as pd
```

```python
# Load the dataset
data = pd.read_csv("students.csv")

# Display the first five records
data.head()
```

```python
# Display column names
print("Original Columns:")
print(data.columns)

# Remove Timestamp column if it exists
if "Timestamp" in data.columns:
    data = data.drop(columns=["Timestamp"])

# Remove completely empty columns
data = data.dropna(axis=1, how="all")

# Remove duplicate records
data = data.drop_duplicates()

# Remove rows with missing values
data = data.dropna()

print("\nDataset cleaned successfully.")

# Display cleaned data
data.head()
```

```python
print("="*60)
print("DATASET SUMMARY")
print("="*60)

print("Total Number of Students:", len(data))

print("\nDepartments")
print(data["Department"].value_counts())

print("\nAverage CGPA")
print(round(data["CGPA"].mean(),2))

print("\nHighest CGPA")
print(data["CGPA"].max())

print("\nLowest CGPA")
print(data["CGPA"].min())
```

```python
for index, row in data.iterrows():

    print("\n" + "="*60)

    print("Student Name:", row["Full Name"])
    print("Department:", row["Department"])
    print("Skills:", row["Skills"])
    print("Programming Languages:", row["Programming Languages Known"])
    print("Career Interest:", row["Career Interest"])
    print("Favourite Course:", row["Favourite Course"])
    print("Career Goal:", row["Career Goal"])
    print("CGPA:", row["CGPA"])
    print("Carryovers:", row["Number of Carryovers"])
```

```python
for index, row in data.iterrows():

    name = row["Full Name"]
    skills = str(row["Skills"]).lower()
    languages = str(row["Programming Languages Known"]).lower()
    interest = str(row["Career Interest"]).lower()

    print("\n")
    print("="*60)
    print("COURSE RECOMMENDATION FOR", name.upper())
    print("="*60)

    print("\nRecommended Courses")

    if "web" in interest:
        print("- HTML")
        print("- CSS")
        print("- JavaScript")
        print("- React")

    elif "backend" in interest:
        print("- Python")
        print("- PHP")
        print("- Node.js")

    elif "data" in interest:
        print("- SQL")
        print("- Machine Learning")
        print("- Data Analysis")

    elif "cyber" in interest:
        print("- Ethical Hacking")
        print("- Network Security")

    elif "mobile" in interest:
        print("- Flutter")
        print("- Kotlin")

    else:
        print("- General Programming")

    print("\nRecommended Certifications")
    print("- Google Career Certificate")
    print("- Microsoft Learn")
    print("- Cisco Networking Academy")
    print("- freeCodeCamp")
```

```python
for index, row in data.iterrows():

    cgpa = float(row["CGPA"])
    carryovers = int(row["Number of Carryovers"])

    print("\nStudent:", row["Full Name"])

    if cgpa >= 4.5:
        print("Performance: Excellent")

    elif cgpa >= 3.5:
        print("Performance: Very Good")

    elif cgpa >= 2.5:
        print("Performance: Good")

    else:
        print("Performance: Needs Improvement")

    if carryovers == 0:
        print("Academic Status: Excellent (No Carryovers)")

    elif carryovers <= 2:
        print("Academic Status: Fair")

    else:
        print("Academic Status: Needs Improvement")
```

```python
required_skills = [
    "python",
    "sql",
    "git",
    "javascript"
]

for index, row in data.iterrows():

    print("\nStudent:", row["Full Name"])

    student_skills = str(row["Skills"]).lower()

    missing = []

    for skill in required_skills:

        if skill not in student_skills:
            missing.append(skill)

    print("Skills to Learn:")

    if len(missing) == 0:
        print("No additional skills recommended.")

    else:
        for skill in missing:
            print("-", skill)
```

```python
recommendations = []

for index, row in data.iterrows():

    cgpa = float(row["CGPA"])

    if cgpa >= 4.5:
        recommendations.append("Artificial Intelligence & Machine Learning")

    elif cgpa >= 3.5:
        recommendations.append("Software Engineering")

    elif cgpa >= 2.5:
        recommendations.append("Web Development")

    else:
        recommendations.append("Programming Fundamentals")

data["Final Recommendation"] = recommendations

print("Recommendations generated successfully.")
```

```python
data.to_csv("student_recommendations.csv", index=False)

print("Recommendation file saved successfully.")

data
```
