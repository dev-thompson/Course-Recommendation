```python
import pandas as pd
```

```python
# Load the dataset
data = pd.read_csv("students.csv")

# Display the first five records
data.head()
```

```python
# Display column names
print("Original Columns:")
print(data.columns)

# Remove Timestamp column if it exists
if "Timestamp" in data.columns:
    data = data.drop(columns=["Timestamp"])

# Remove completely empty columns
data = data.dropna(axis=1, how="all")

# Remove duplicate records
data = data.drop_duplicates()

# Remove rows with missing values
data = data.dropna()

print("\nDataset cleaned successfully.")

# Display cleaned data
data.head()
```

```python
for index, row in data.iterrows():

    cgpa = float(row["CGPA"])
    carryovers = int(row["Number of Carryovers"])

    print("\nStudent:", row["Full Name"])

    if cgpa >= 4.5:
        print("Performance: Excellent")

    elif cgpa >= 3.5:
        print("Performance: Very Good")

    elif cgpa >= 2.5:
        print("Performance: Good")

    else:
        print("Performance: Needs Improvement")

    if carryovers == 0:
        print("Academic Status: Excellent (No Carryovers)")

    elif carryovers <= 2:
        print("Academic Status: Fair")

    else:
        print("Academic Status: Needs Improvement")
```

```python
required_skills = [
    "python",
    "sql",
    "git",
    "javascript"
]

for index, row in data.iterrows():

    print("\nStudent:", row["Full Name"])

    student_skills = str(row["Skills"]).lower()

    missing = []

    for skill in required_skills:

        if skill not in student_skills:
            missing.append(skill)

    print("Skills to Learn:")

    if len(missing) == 0:
        print("No additional skills recommended.")

    else:
        for skill in missing:
            print("-", skill)
```

```python
recommendations = []

for index, row in data.iterrows():

    cgpa = float(row["CGPA"])

    if cgpa >= 4.5:
        recommendations.append("Artificial Intelligence & Machine Learning")

    elif cgpa >= 3.5:
        recommendations.append("Software Engineering")

    elif cgpa >= 2.5:
        recommendations.append("Web Development")

    else:
        recommendations.append("Programming Fundamentals")

data["Final Recommendation"] = recommendations

print("Recommendations generated successfully.")
```

```python
course_mapping = {
    "ai": "artificial intelligence",
    "artificial intelligence": "artificial intelligence",

    "front end development": "frontend development",
    "front-end development": "frontend development",
    "frontend development": "frontend development",

    "data base": "database",
    "database": "database",

    "cyber security": "cybersecurity",
    "cyber-security": "cybersecurity",

    "ui,/ux design": "ui/ux design",
    "ui/ux": "ui/ux design",
    "ui/ux design": "ui/ux design"
}

df["Favourite Course"] = (
    df["Favourite Course"]
      .replace(course_mapping)
)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
tfidf = TfidfVectorizer(stop_words='english')

tfidf_matrix = tfidf.fit_transform(df['Student_Profile'])

print(tfidf_matrix.shape)
```

```python
cosine_sim = cosine_similarity(tfidf_matrix)

print(cosine_sim.shape)
```

```python
def recommend_courses(student_name, top_n=10):

    # Check if student exists
    if student_name not in df['Full Name'].values:
        return "Student not found."

    # Student index
    idx = df[df['Full Name'] == student_name].index[0]

    # Student's favourite course
    current_course = df.iloc[idx]['Favourite Course']

    # Similarity scores
    similarity_scores = list(enumerate(cosine_sim[idx]))
    similarity_scores = sorted(similarity_scores,
                               key=lambda x: x[1],
                               reverse=True)

    # Ignore the student
    similarity_scores = similarity_scores[1:]

    recommended_courses = []

    print("="*60)
    print("Student:", student_name)
    print("Current Favourite Course:", current_course)
    print("="*60)

    # Look at the top similar students
    for i, score in similarity_scores[:top_n]:

        course = df.iloc[i]['Favourite Course']

        # Skip if it's the student's own course
        if course.lower() != current_course.lower():
            recommended_courses.append(course)

    recommendations = (
        pd.Series(recommended_courses)
          .value_counts()
          .head(5)
    )

    print("\nRecommended Courses\n")
    print(recommendations)

    return recommendations
```

```python
def cgpa_category(cgpa):
    if cgpa >= 4.5:
        return "excellent_cgpa"
    elif cgpa >= 3.5:
        return "good_cgpa"
    elif cgpa >= 2.5:
        return "average_cgpa"
    else:
        return "low_cgpa"


df["CGPA_Category"] = df["CGPA"].apply(cgpa_category)
```
