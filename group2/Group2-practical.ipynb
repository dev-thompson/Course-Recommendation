```python
import pandas as pd
```

```python
data = pd.read_csv("students.csv")

data.head()
```

```python
for index, row in data.iterrows():

    full_name = row["Full Name"]
    department = row["Department"]
    skills = str(row["Skills"]).lower().split(",")
    languages = str(row["Programming Languages Known"]).lower().split(",")
    interest = str(row["Career Interest"]).lower()
    favourite_course = row["Favourite Course"]
    career_goal = row["Career Goal"]
    cgpa = float(row["CGPA"])
    carryovers = int(row["Number of Carryovers"])

    print("="*50)
    print("Name:", full_name)
    print("Department:", department)
    print("Skills:", skills)
    print("Languages:", languages)
    print("Career Interest:", interest)
    print("Favourite Course:", favourite_course)
    print("Career Goal:", career_goal)
    print("CGPA:", cgpa)
    print("Carryovers:", carryovers)
```

```python
for index, row in data.iterrows():

    full_name = row["Full Name"]
    languages = str(row["Programming Languages Known"]).lower().split(",")
    interest = str(row["Career Interest"]).lower()

    print("\n")
    print("=" * 60)
    print("Recommendations for:", full_name)
    print("=" * 60)

    # Recommended Courses
    print("\nRecommended Courses:")

    if "web" in interest:
        print("- HTML & CSS")
        print("- JavaScript")
        print("- React")

    elif "backend" in interest:
        print("- Python")
        print("- PHP")
        print("- Node.js")

    elif "data" in interest:
        print("- Python for Data Science")
        print("- SQL")
        print("- Machine Learning")

    elif "cyber" in interest:
        print("- Ethical Hacking")
        print("- Linux")

    else:
        print("- General Programming")

    # Certifications
    print("\nRecommended Certifications:")
    print("- Google Career Certificate")
    print("- freeCodeCamp")
    print("- Microsoft Learn")
    print("- Alison")

    # Career Path
    print("\nRecommended Career Path:")

    if "web" in interest:
        print("- Frontend Developer")

    elif "backend" in interest:
        print("- Backend Developer")

    elif "data" in interest:
        print("- Data Scientist")

    elif "cyber" in interest:
        print("- Cybersecurity Analyst")

    else:
        print("- Software Developer")

    # Programming Languages to Learn
    print("\nProgramming Languages to Learn:")

    if "python" not in languages:
        print("- Python")

    if "javascript" not in languages:
        print("- JavaScript")

    if "java" not in languages:
        print("- Java")

    if "c++" not in languages:
        print("- C++")
```
